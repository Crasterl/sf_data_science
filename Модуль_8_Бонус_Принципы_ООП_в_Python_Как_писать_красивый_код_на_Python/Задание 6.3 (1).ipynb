Определите класс `Dog`, у которого есть два метода:

1. `bark`,
2. `give_paw`.

Оба метода должны принимать любое количество аргументов. Метод `bark` должен возвращать строку `"Bark!"`, а метод `give_paw` — строку `"Paw"`.

В ячейке ниже дана заготовка для написания кода.


**Примеры работы программы:**

```python
dog = Dog()

print(dog.bark())
Bark!

print(dog.give_paw())
Paw

print(dog.bark(1, 2, 3))
Bark!

print(dog.give_paw("hello", True, 42))
Paw
```

````python
def test_1():
    dog = Dog()
    assert dog.bark() == "Bark!", "bark() должен вернуть 'Bark!'"


def test_2():
    dog = Dog()
    assert dog.give_paw() == "Paw", "give_paw() должен вернуть 'Paw'"


def test_3():
    dog = Dog()
    assert dog.bark(1, 2, 3) == "Bark!", \
        "bark() должен принимать несколько аргументов"


def test_4():
    dog = Dog()
    assert dog.give_paw("hello", True, 42) == "Paw", \
        "give_paw() должен принимать несколько аргументов"


def test_5():
    dog = Dog()
    assert dog.bark("a", "b", "c", "d", "e") == "Bark!", \
        "bark() должен работать с любым количеством аргументов"


def test_6():
    dog = Dog()
    assert dog.give_paw() == "Paw", \
        "give_paw() без аргументов должен вернуть 'Paw'"


def test_7():
    dog = Dog()
    assert dog.bark(None, [1, 2], {"key": "value"}) == "Bark!", \
        "bark() должен работать с аргументами разных типов"


tests = [
    ("test_1", test_1),
    ("test_2", test_2),
    ("test_3", test_3),
    ("test_4", test_4),
    ("test_5", test_5),
    ("test_6", test_6),
    ("test_7", test_7),
]

passed = 0

for test_name, test_func in tests:
    try:
        test_func()
        print(f"✅ {test_name}: тест пройден")
        passed += 1
    except AssertionError as error:
        print(f"❌ {test_name}: тест не пройден")
        print(f"   {error}")
    except Exception as error:
        print(f"❌ {test_name}: ошибка выполнения")
        print(f"   {type(error).__name__}: {error}")

print()

if passed == len(tests):
    print("Проверка завершена! Все тесты пройдены.")
else:
    print(f"Проверка завершена. Пройдено: {passed} из {len(tests)}.")
````

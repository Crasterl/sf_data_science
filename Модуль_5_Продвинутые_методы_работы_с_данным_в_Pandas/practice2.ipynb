```python
import pandas as pd
```

```python
orders = pd.read_csv('data/orders.csv', sep=';')
products = pd.read_csv('data/products.csv', sep=';')
```

```python
orders_products = orders.merge(right=products,
                               how='left',
                               left_on='ID товара',
                               right_on='Product_ID',

                               )
orders_products.head()
```

```python
mask1 = orders_products['Product_ID'].isnull()
```

```python
orders_products[mask1]
```

```python
mask1 = orders_products['Отменен'] == 'Да'
orders_products[mask1]
```

```python
mask1 = orders_products['Оплачен'] == 'Да'
mask2 = orders_products['Product_ID'].notnull()
orders_products['future_profit'] = orders_products['Количество'] * orders_products['Price']
orders_products[mask1 & mask2].groupby(by='ID Покупателя')['future_profit'].sum()
```
